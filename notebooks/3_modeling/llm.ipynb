```python

# =========================================================
# CELDA 1 â€” Imports + configuraciÃ³n LLM
# Nuevo notebook: 07_llm_features.ipynb
# =========================================================
# pip install groq          â† Groq API (gratuita, rÃ¡pida)
# pip install ollama        â† Ollama local (privacidad total)
# =========================================================

import os, re, json, time, warnings
import numpy as np
import pandas as pd
from pathlib import Path
from collections import Counter
from scipy.optimize import minimize
from scipy.stats import kruskal
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.metrics import f1_score, classification_report
import lightgbm as lgb
from lightgbm import LGBMClassifier

warnings.filterwarnings("ignore")

# â”€â”€ Proyecto â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
from triaje_ia.config import DATA_PROCESSED, MODELS_DIR
from triaje_ia.data.cleaner import cargar_dataset_limpio

NOTEBOOK_DIR = Path.cwd()

# â”€â”€ Helpers (idÃ©nticos a prueba_v2) â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
def macro_f1(y_true, y_pred):
    return f1_score(y_true, y_pred, average="macro", zero_division=0)

def aplicar_thresholds(proba, w):
    return np.argmax(proba * w, axis=1)

def objetivo_th(w, proba, y_true):
    return -f1_score(y_true, aplicar_thresholds(proba, w),
                     average="macro", zero_division=0)

# â”€â”€ Baseline oficial del proyecto â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# Estos son los resultados que el LLM tiene que superar:
F1_REFERENCIA = {
    "exp5b_th_test"    : 0.5092,   # sin NLP
    "exp5b_nlp_th_test": 0.5615,   # + TF-IDF + TH  â† mÃ­nimo a batir
    "exp5b_bert_cv"    : 0.5743,   # + BERT-SVD  â† mejor CV conocido
}

# â”€â”€ ConfiguraciÃ³n LLM â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
LLM_BACKEND  = os.getenv("LLM_BACKEND", "groq")   # "groq" | "ollama"
GROQ_API_KEY = os.getenv("GROQ_API_KEY", "")       # definir en .env
GROQ_MODEL   = "llama-3.1-8b-instant"              # gratuito, ~800 tok/s
OLLAMA_MODEL = "llama3.1:8b"

if LLM_BACKEND == "groq":
    from groq import Groq
    _client = Groq(api_key=GROQ_API_KEY)

    def llamar_llm(prompt: str, system: str = "",
                   temperature: float = 0.0, max_tokens: int = 256) -> str:
        msgs = []
        if system:
            msgs.append({"role": "system", "content": system})
        msgs.append({"role": "user", "content": prompt})
        r = _client.chat.completions.create(
            model=GROQ_MODEL, messages=msgs,
            temperature=temperature, max_tokens=max_tokens,
        )
        return r.choices[0].message.content.strip()

elif LLM_BACKEND == "ollama":
    import requests

    def llamar_llm(prompt: str, system: str = "",
                   temperature: float = 0.0, max_tokens: int = 256) -> str:
        r = requests.post("http://localhost:11434/api/generate", json={
            "model": OLLAMA_MODEL, "prompt": prompt, "system": system,
            "stream": False,
            "options": {"temperature": temperature, "num_predict": max_tokens},
        }, timeout=60)
        r.raise_for_status()
        return r.json()["response"].strip()

# â”€â”€ Test rÃ¡pido de conexiÃ³n â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
try:
    ping = llamar_llm("Responde solo: OK", max_tokens=5)
    print(f"âœ… LLM '{LLM_BACKEND}' listo  â†’  '{ping}'")
except Exception as e:
    print(f"âŒ Error LLM: {e}")

# â”€â”€ Cargar artefactos base â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
X_train = pd.read_parquet(DATA_PROCESSED / "X_train.parquet")
X_test  = pd.read_parquet(DATA_PROCESSED / "X_test.parquet")
y_train = pd.read_parquet(DATA_PROCESSED / "y_train.parquet").squeeze()
y_test  = pd.read_parquet(DATA_PROCESSED / "y_test.parquet").squeeze()

groups_train         = np.load(DATA_PROCESSED / "groups_train.npy")
sample_weights_train = np.load(DATA_PROCESSED / "sample_weights_train.npy")

with open(MODELS_DIR / "lgbm_best_params.json", encoding="utf-8") as f:
    params_modelo = {k: v for k, v in json.load(f).items()
                     if not k.startswith("_")}

CV = StratifiedGroupKFold(n_splits=5, shuffle=False)

# â”€â”€ Cargar exp5b + BERT (mejor punto de partida conocido) â”€
X_tr_exp5b = pd.read_parquet(DATA_PROCESSED / "X_train_exp5b.parquet")
X_te_exp5b = pd.read_parquet(DATA_PROCESSED / "X_test_exp5b.parquet")

df_bert_train = pd.read_parquet(DATA_PROCESSED / "bert_embeddings_train.parquet")
df_bert_test  = pd.read_parquet(DATA_PROCESSED / "bert_embeddings_test.parquet")

# BERT-SVD: todos los componentes (15) â€” ya validados en prueba_v2
X_tr_bert = pd.concat([
    X_tr_exp5b.reset_index(drop=True),
    df_bert_train.reset_index(drop=True),
], axis=1)
X_te_bert = pd.concat([
    X_te_exp5b.reset_index(drop=True),
    df_bert_test.reset_index(drop=True),
], axis=1)

print(f"\nX_tr_bert : {X_tr_bert.shape}   â† punto de partida")
print(f"y_train   : {y_train.value_counts().sort_index().to_dict()}")
print(f"\nReferencias a batir:")
for k, v in F1_REFERENCIA.items():
    print(f"  {k:<28} {v:.4f}")
print("\nâœ… CELDA 1 OK")
```

```python
# =========================================================
# CELDA 2 â€” RadiografÃ­a del chiefcomplaint en acuity 4 y 5
# Objetivo: entender quÃ© patrones existen antes de diseÃ±ar
# el prompt del LLM. Sin este anÃ¡lisis el prompt es ciego.
# =========================================================

df = cargar_dataset_limpio(forzar=False)

fecha_corte = df["intime"].quantile(0.80)
df_train_raw = df[df["intime"] <= fecha_corte].reset_index(drop=True)
df_test_raw  = df[df["intime"] >  fecha_corte].reset_index(drop=True)

assert len(df_train_raw) == len(X_train), "DesalineaciÃ³n df_train_raw"

# â”€â”€ Preprocesado mÃ­nimo â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
def limpiar(t):
    if pd.isna(t): return ""
    return str(t).lower().strip()

df_train_raw["cc"] = df_train_raw["chiefcomplaint"].apply(limpiar)

STOPWORDS = {"the","a","an","and","or","for","of","to","with","in","on",
             "at","by","is","are","was","be","has","have","patient","pt","s"}

def top_palabras(serie, n=30):
    c = Counter()
    for t in serie.dropna():
        c.update(w for w in re.findall(r"\b[a-z]{3,}\b", t) if w not in STOPWORDS)
    return c.most_common(n)

# â”€â”€ DistribuciÃ³n general â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("=" * 68)
print("  DISTRIBUCIÃ“N ACUITY â€” dataset train")
print("=" * 68)
for nivel, n in df_train_raw["acuity"].value_counts().sort_index().items():
    pct = n / len(df_train_raw) * 100
    print(f"  Acuity {nivel}  {n:>8,}  ({pct:5.2f}%)  {'â–ˆ'*int(pct/2)}")

# â”€â”€ AnÃ¡lisis por nivel â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
for nivel in [5, 4]:
    sub = df_train_raw[df_train_raw["acuity"] == nivel].copy()
    print(f"\n{'='*68}")
    print(f"  ACUITY {nivel}  â€”  {len(sub):,} casos  ({len(sub)/len(df_train_raw)*100:.2f}%)")
    print(f"{'='*68}")

    # Top 40 chief complaints exactos
    top_cc = sub["chiefcomplaint"].str.lower().str.strip().value_counts().head(40)
    print(f"\n  â–¸ TOP 40 chief complaints (texto exacto):")
    for txt, cnt in top_cc.items():
        print(f"    {cnt:>4,}  ({cnt/len(sub)*100:4.1f}%)  {txt!r}")

    # Top palabras
    print(f"\n  â–¸ Top 30 palabras clave:")
    for palabra, cnt in top_palabras(sub["cc"], 30):
        print(f"    {cnt:>5,}  {palabra}")

    # DistribuciÃ³n pain score
    if "pain" in sub.columns:
        print(f"\n  â–¸ Pain score:")
        for v, c in sub["pain"].value_counts(dropna=False).sort_index().items():
            label = "NaN" if pd.isna(v) else str(int(v))
            print(f"    pain={label:<3}  {c:>5,}  ({c/len(sub)*100:.1f}%)")

    # Vitales: zona_verde si existe en X_train
    if "zona_verde" in X_train.columns:
        zv = X_train.loc[df_train_raw["acuity"] == nivel, "zona_verde"].mean()
        print(f"\n  â–¸ % zona_verde : {zv*100:.1f}%")

    # 30 ejemplos aleatorios (texto original, sin procesar)
    print(f"\n  â–¸ 30 ejemplos aleatorios (texto original):")
    muestra = sub["chiefcomplaint"].dropna().sample(
        min(30, len(sub)), random_state=42).tolist()
    for i, ej in enumerate(muestra, 1):
        print(f"    {i:>2}. {ej!r}")

    # Longitud de texto
    lng = sub["cc"].str.len()
    print(f"\n  â–¸ Longitud texto â€” media: {lng.mean():.1f} | "
          f"mediana: {lng.median():.0f} | max: {lng.max()}")

# â”€â”€ Comparativa directa: quÃ© separa el 4 del 5 â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print(f"\n{'='*68}")
print("  DIFERENCIACIÃ“N 4 vs 5 â€” tÃ©rminos exclusivos (top 60 cada clase)")
print(f"{'='*68}")

sub4 = df_train_raw[df_train_raw["acuity"] == 4]
sub5 = df_train_raw[df_train_raw["acuity"] == 5]

set_cc4 = set(sub4["chiefcomplaint"].str.lower().str.strip()
              .value_counts().head(60).index)
set_cc5 = set(sub5["chiefcomplaint"].str.lower().str.strip()
              .value_counts().head(60).index)

solo_5 = set_cc5 - set_cc4
solo_4 = set_cc4 - set_cc5

vc5 = sub5["chiefcomplaint"].str.lower().str.strip().value_counts()
vc4 = sub4["chiefcomplaint"].str.lower().str.strip().value_counts()

print("\n  Solo en Acuity 5 (â‰ˆ trivial / administrativo):")
for cc_txt in sorted(solo_5, key=lambda x: -vc5.get(x, 0)):
    print(f"    {vc5.get(cc_txt,0):>4,}  {cc_txt!r}")

print("\n  Solo en Acuity 4 (â‰ˆ urgente moderado):")
for cc_txt in sorted(solo_4, key=lambda x: -vc4.get(x, 0)):
    print(f"    {vc4.get(cc_txt,0):>4,}  {cc_txt!r}")

# â”€â”€ Palabras DIFERENCIALES (TF-IDF manual) â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print(f"\n{'='*68}")
print("  PALABRAS CON MAYOR RATIO 5/4 y 4/5 (sin stopwords)")
print(f"{'='*68}")

c5 = Counter(w for t in sub5["cc"].dropna()
             for w in re.findall(r"\b[a-z]{3,}\b", t) if w not in STOPWORDS)
c4 = Counter(w for t in sub4["cc"].dropna()
             for w in re.findall(r"\b[a-z]{3,}\b", t) if w not in STOPWORDS)

total5 = sum(c5.values()) + 1
total4 = sum(c4.values()) + 1

# Palabras que aparecen en ambas con suficiente frecuencia
palabras_comunes = set(c5.keys()) & set(c4.keys())
palabras_comunes = {p for p in palabras_comunes if c5[p] + c4[p] > 20}

ratios = {p: (c5[p]/total5) / (c4[p]/total4 + 1e-9)
          for p in palabras_comunes}

print("\n  Palabras MÃS propias de Acuity 5 (ratio 5/4 alto):")
for p, r in sorted(ratios.items(), key=lambda x: -x[1])[:20]:
    print(f"    ratio={r:5.2f}  acuity5={c5[p]:>4,}  acuity4={c4[p]:>4,}  '{p}'")

print("\n  Palabras MÃS propias de Acuity 4 (ratio 4/5 alto):")
for p, r in sorted(ratios.items(), key=lambda x: x[1])[:20]:
    print(f"    ratio={1/r:5.2f}  acuity4={c4[p]:>4,}  acuity5={c5[p]:>4,}  '{p}'")

print(f"\n{'='*68}")
print("  âœ… CELDA 2 completada. Pega el output para diseÃ±ar el prompt.")
print(f"{'='*68}")
```

```python
# =========================================================
# CELDA 3 â€” Prompt calibrado con los datos reales de CELDA 2
# =========================================================

SYSTEM_PROMPT = """
Eres un sistema experto en triaje de urgencias hospitalarias.
RecibirÃ¡s un motivo de consulta escrito por una enfermera, en espaÃ±ol o inglÃ©s.
Devuelve ÃšNICAMENTE un objeto JSON vÃ¡lido, sin explicaciones, sin markdown, sin texto adicional.

â”â”â” GUÃA DE CLASIFICACIÃ“N (aprendida de datos reales) â”â”â”

SEÃ‘ALES DE ACUITY 5 / NO URGENTE (trÃ¡mite administrativo o consulta trivial):
  - Retirada de puntos o grapas: "suture removal", "staples removal", "retirada de puntos"
  - Recarga de medicaciÃ³n habitual: "med refill", "medication refill", "needs rx", "requesting rx", "renovar receta"
  - AnalÃ­tica o extracciÃ³n de sangre: "lab draw", "lab request", "labs only", "analÃ­tica de control"
  - VacunaciÃ³n rutinaria de seguimiento: "rabies vaccine", "tetanus booster", "vacuna antitetÃ¡nica"
  - RevisiÃ³n de herida en seguimiento SIN sÃ­ntomas nuevos: "wound check", "wound eval" (solo si no hay signos de infecciÃ³n ni dolor activo)
  - TrÃ¡mites sociales o administrativos: "social work consult", "medical clearance", "aclaramiento mÃ©dico"
  - ExposiciÃ³n animal sin sÃ­ntomas: "bat exposure", "tick bite"
  - Enrojecimiento ocular aislado sin dolor: "eye redness", "ojo rojo sin dolor"

SEÃ‘ALES DE ACUITY 4 / MENOS URGENTE (requiere evaluaciÃ³n clÃ­nica real):
  - Cualquier CAÃDA reciente: "s/p fall", "trip and fall", "caÃ­da en domicilio"
  - Accidente de trÃ¡fico: "mvc", "accidente de trÃ¡fico", "RTA"
  - Traumatismo con localizaciÃ³n especÃ­fica: tobillo, rodilla, muÃ±eca, cuello, hombro, cadera, pierna
  - LaceraciÃ³n activa o herida con sangrado: "laceration", "lac", "finger lac", "herida incisa"
  - SÃ­ntomas sistÃ©micos: tos, disuria, epistaxis, cefalea, ansiedad, hematuria, dolor rectal, faringitis
  - IntoxicaciÃ³n: "etoh", "alcohol intoxication"
  - AgresiÃ³n: "assault", "human bite", "agresiÃ³n"
  - "wound eval" CON signos de infecciÃ³n o dolor activo â†’ acuity 4, NO 5
  - "suture removal" CON signos de infecciÃ³n â†’ acuity 4, NO 5

CASOS AMBIGUOS â€” regla de oro:
  El mismo motivo puede ser acuity 4 o 5 segÃºn el contexto.
  "back pain" aislado sin trauma â†’ puede ser 4 o 5.
  "wound eval" + infecciÃ³n â†’ 4. "wound eval" + seguimiento rutinario â†’ 5.
  Si hay duda, prioriza la seguridad del paciente â†’ asigna el nivel mÃ¡s urgente.

â”â”â” VARIABLES A EXTRAER â”â”â”

"es_tramite_administrativo" (0 o 1):
  1 = puramente logÃ­stico/administrativo sin urgencia mÃ©dica activa.
  Palabras fuertemente asociadas: refill, removal, draw, lab, requesting, needs rx, staples, vaccine, booster.
  0 = cualquier sÃ­ntoma activo, dolor nuevo, trauma, infecciÃ³n.

"es_trauma_agudo" (0 o 1):
  1 = hay un evento traumÃ¡tico reciente (caÃ­da, MVC, golpe, dislocaciÃ³n, fractura sospechada, agresiÃ³n).

"es_dolor_agudo_intenso" (0 o 1):
  1 = dolor agudo o intenso NO crÃ³nico. Incluye: laceraciÃ³n activa, absceso, cefalea intensa, dolor torÃ¡cico, cÃ³lico.

"es_sintoma_sistemico" (0 o 1):
  1 = sÃ­ntoma que afecta estado general: fiebre, tos, disuria, epistaxis, hematuria, disnea, vÃ³mitos, mareo, palpitaciones.

"es_lesion_menor_localizada" (0 o 1):
  1 = lesiÃ³n fÃ­sica menor y localizada sin compromiso sistÃ©mico (dedo cortado, ojo rojo, picadura, Ãºlcera pequeÃ±a estable).

"menciona_parte_cuerpo" (0 o 1):
  1 = menciona explÃ­citamente una parte del cuerpo (rodilla, tobillo, dedo, espalda, ojo, oÃ­do, hombro, muÃ±eca, cadera, etc.).

"es_infeccion_sospechada" (0 o 1):
  1 = signos o sospecha de infecciÃ³n: celulitis, absceso, herida infectada, ojo con secreciÃ³n, faringitis, ITU.

"es_dolor_cronico_conocido" (0 o 1):
  1 = dolor claramente crÃ³nico sin evento agudo desencadenante (ej: "lower back pain" sin trauma, "dental pain" de seguimiento).

"urgencia_llm" (entero 1 a 5):
  Nivel ESI estimado.
  5 = trÃ¡mite administrativo o sÃ­ntoma completamente trivial.
  4 = evaluaciÃ³n clÃ­nica necesaria pero no urgente (dolor moderado, trauma menor, sÃ­ntoma localizado).
  3 = mÃºltiples quejas, dolor agudo importante, necesita recursos.
  2 = emergente, riesgo de deterioro.
  1 = resucitaciÃ³n inmediata.

Devuelve SOLO el JSON. Formato exacto:
{"es_tramite_administrativo":0,"es_trauma_agudo":0,"es_dolor_agudo_intenso":0,"es_sintoma_sistemico":0,"es_lesion_menor_localizada":0,"menciona_parte_cuerpo":0,"es_infeccion_sospechada":0,"es_dolor_cronico_conocido":0,"urgencia_llm":3}
""".strip()

USER_TEMPLATE = "Motivo de consulta: \"{cc}\""

# â”€â”€ FunciÃ³n de parseo robusta â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
CAMPOS_ESPERADOS = [
    "es_tramite_administrativo",
    "es_trauma_agudo",
    "es_dolor_agudo_intenso",
    "es_sintoma_sistemico",
    "es_lesion_menor_localizada",
    "menciona_parte_cuerpo",
    "es_infeccion_sospechada",
    "es_dolor_cronico_conocido",   # â† nueva vs versiÃ³n anterior
    "urgencia_llm",
]
VALORES_DEFECTO = {c: -1 for c in CAMPOS_ESPERADOS}

def parsear_respuesta(texto: str) -> dict:
    match = re.search(r"\{[^{}]+\}", texto, re.DOTALL)
    if not match:
        return VALORES_DEFECTO.copy()
    try:
        d = json.loads(match.group())
        resultado = VALORES_DEFECTO.copy()
        for k in CAMPOS_ESPERADOS:
            if k in d:
                resultado[k] = int(d[k])
        return resultado
    except Exception:
        return VALORES_DEFECTO.copy()

# â”€â”€ Test con ejemplos que cubren los casos AMBIGUOS detectados â”€â”€
EJEMPLOS_TEST = [
    # --- Acuity 5 claros ---
    ("EN_a5_claro",   "SUTURE REMOVAL"),
    ("EN_a5_claro",   "Med refill"),
    ("EN_a5_claro",   "lab draw"),
    ("EN_a5_claro",   "rabies vaccine"),
    ("EN_a5_claro",   "wound check"),            # seguimiento rutinario
    # --- Acuity 4 claros ---
    ("EN_a4_claro",   "S/P TRIP AND FALL"),
    ("EN_a4_claro",   "R SHOULDER DISLOCATION"),
    ("EN_a4_claro",   "Dysuria, Lower abdominal pain"),
    ("EN_a4_claro",   "EPISTAXIS"),
    ("EN_a4_claro",   "ETOH"),                   # intoxicaciÃ³n etÃ­lica
    # --- AMBIGUOS detectados en CELDA 2 ---
    ("EN_ambig",      "wound eval"),             # puede ser 4 o 5 â†’ Â¿quÃ© decide el LLM?
    ("EN_ambig",      "lower back pain"),        # ratio 4/5 = 3.40 â†’ tiende a 4
    ("EN_ambig",      "back pain"),              # aparece en ambos acuities
    ("EN_ambig",      "INFECTED RIGHT EARLOBE"), # infecciÃ³n â†’ debe ir a 4
    ("EN_ambig",      "CELLULITIS"),             # infecciÃ³n â†’ debe ir a 4, no 5
    # --- EspaÃ±ol producciÃ³n ---
    ("ES_a5",         "retirada de puntos en herida de rodilla"),
    ("ES_a5",         "renovaciÃ³n de receta para tensiÃ³n"),
    ("ES_a5",         "analÃ­tica de control, sin sÃ­ntomas"),
    ("ES_a4",         "caÃ­da en domicilio, dolor en tobillo derecho"),
    ("ES_a4",         "paciente refiere tos y fiebre desde hace 3 dÃ­as"),
]

print("=" * 80)
print("  TEST DEL PROMPT v2 â€” calibrado con datos reales de CELDA 2")
print("=" * 80)
print(f"{'ETIQUETA':<18} {'urg':>3} {'adm':>3} {'trm':>3} {'dlr':>3} "
      f"{'sis':>3} {'men':>3} {'cpo':>3} {'inf':>3} {'cro':>3}  TEXTO")
print("-" * 100)

resultados_test = []
errores = 0

for etiqueta, cc in EJEMPLOS_TEST:
    try:
        raw = llamar_llm(
            prompt=USER_TEMPLATE.format(cc=cc),
            system=SYSTEM_PROMPT,
            temperature=0.0,
            max_tokens=150,
        )
        parsed = parsear_respuesta(raw)
    except Exception as ex:
        parsed = VALORES_DEFECTO.copy()
        errores += 1

    resultados_test.append({"etiqueta": etiqueta, "cc": cc, **parsed})

    print(
        f"{etiqueta:<18} "
        f"{parsed['urgencia_llm']:>3} "
        f"{parsed['es_tramite_administrativo']:>3} "
        f"{parsed['es_trauma_agudo']:>3} "
        f"{parsed['es_dolor_agudo_intenso']:>3} "
        f"{parsed['es_sintoma_sistemico']:>3} "
        f"{parsed['es_lesion_menor_localizada']:>3} "
        f"{parsed['menciona_parte_cuerpo']:>3} "
        f"{parsed['es_infeccion_sospechada']:>3} "
        f"{parsed['es_dolor_cronico_conocido']:>3} "
        f"  {cc[:45]}"
    )

df_test_prompt = pd.DataFrame(resultados_test)

# â”€â”€ ValidaciÃ³n por categorÃ­a â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("-" * 100)
print("\n  CHECKS AUTOMÃTICOS:")

checks = {
    "a5_claro â†’ es_tramite_administrativo=1":
        (df_test_prompt[df_test_prompt["etiqueta"]=="EN_a5_claro"]["es_tramite_administrativo"] == 1).all(),
    "a4_claro â†’ es_tramite_administrativo=0":
        (df_test_prompt[df_test_prompt["etiqueta"]=="EN_a4_claro"]["es_tramite_administrativo"] == 0).all(),
    "INFECTED EARLOBE â†’ es_infeccion_sospechada=1":
        (df_test_prompt[df_test_prompt["cc"]=="INFECTED RIGHT EARLOBE"]["es_infeccion_sospechada"] == 1).any(),
    "CELLULITIS â†’ es_infeccion_sospechada=1":
        (df_test_prompt[df_test_prompt["cc"]=="CELLULITIS"]["es_infeccion_sospechada"] == 1).any(),
    "S/P TRIP AND FALL â†’ es_trauma_agudo=1":
        (df_test_prompt[df_test_prompt["cc"]=="S/P TRIP AND FALL"]["es_trauma_agudo"] == 1).any(),
    "ES a5 â†’ es_tramite_administrativo=1":
        (df_test_prompt[df_test_prompt["etiqueta"]=="ES_a5"]["es_tramite_administrativo"] == 1).all(),
    "ES a4 â†’ es_tramite_administrativo=0":
        (df_test_prompt[df_test_prompt["etiqueta"]=="ES_a4"]["es_tramite_administrativo"] == 0).all(),
}

for desc, ok in checks.items():
    estado = "âœ…" if ok else "âŒ FALLO"
    print(f"  {estado}  {desc}")

n_ok = sum(checks.values())
print(f"\n  Resultado: {n_ok}/{len(checks)} checks pasados")
print(f"  â†’ Si â‰¥ 6/7: prompt listo para batch en CELDA 4")
print(f"  â†’ Si < 5/7: revisar los ejemplos que fallan y ajustar la guÃ­a del prompt")
print(f"\nâœ… CELDA 3 OK")
```

```python
# =========================================================
# CELDA 4 â€” Batch OPTIMIZADO (Solo valores Ãºnicos)
# Groq procesarÃ¡ ~50k filas en lugar de 418k (Ahorro del 88%)
# =========================================================

import time
import pandas as pd
from tenacity import retry, wait_exponential, stop_after_attempt

CACHE_PATH  = DATA_PROCESSED / "llm_features_dict_cache.parquet"
BATCH_SIZE  = 100
PAUSA_REQ   = 0.6
PAUSA_LOTE  = 2.0

# â”€â”€ 1. Construir dataset global y extraer ÃšNICOS â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
df_all = pd.concat([
    df_train_raw[["chiefcomplaint"]].assign(split="train"),
    df_test_raw[["chiefcomplaint"]].assign(split="test"),
], ignore_index=True)

df_all["idx"] = df_all.index.astype(str)
df_all["cc_clean"] = df_all["chiefcomplaint"].fillna("").str.strip().str.lower() # Pasamos a minÃºsculas para unificar mÃ¡s

# Â¡EL ATAJO! Sacamos solo los textos Ãºnicos
textos_unicos = df_all["cc_clean"].drop_duplicates().reset_index(drop=True)
df_unicos = pd.DataFrame({"cc_clean": textos_unicos})

print(f"  ðŸ“Š Filas totales: {len(df_all):,}")
print(f"  ðŸ”¥ Textos Ãºnicos a procesar: {len(df_unicos):,} (Ahorro del {(1 - len(df_unicos)/len(df_all))*100:.1f}%)")

# â”€â”€ 2. GestiÃ³n de CachÃ© â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
if CACHE_PATH.exists():
    df_cache = pd.read_parquet(CACHE_PATH)
    ya_procesados = set(df_cache["cc_clean"])
    print(f"  ðŸ“‚ CachÃ© encontrada: {len(df_cache):,} textos ya procesados.")
else:
    df_cache = pd.DataFrame()
    ya_procesados = set()

pendientes = df_unicos[~df_unicos["cc_clean"].isin(ya_procesados)].copy()
print(f"  ðŸ“‹ Textos pendientes: {len(pendientes):,} | Tiempo estimado: ~{(len(pendientes)*PAUSA_REQ)/60:.1f} min")

# â”€â”€ 3. Funciones del LLM (Sin cambios) â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
def error_defecto(retry_state):
    return VALORES_DEFECTO.copy()

@retry(
    wait=wait_exponential(multiplier=2, min=2, max=30), 
    stop=stop_after_attempt(5),
    retry_error_callback=error_defecto
)
def llamar_llm_seguro(cc_texto: str) -> dict:
    raw = llamar_llm(
        prompt=USER_TEMPLATE.format(cc=cc_texto),
        system=SYSTEM_PROMPT,
        temperature=0.0,
        max_tokens=128,
    )
    parsed = parsear_respuesta(raw)
    if all(v == -1 for v in parsed.values()):
        raise ValueError("Parse fallido")
    return parsed

# â”€â”€ 4. Bucle de Procesamiento â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
nuevas_filas = []
t_inicio = time.time()
n_errores = 0

textos_pendientes = pendientes["cc_clean"].tolist()

if len(textos_pendientes) > 0:
    print(f"\n{'='*68}")
    print(f"  PROCESANDO {len(textos_pendientes):,} TEXTOS ÃšNICOS")
    print(f"{'='*68}")

    for i in range(0, len(textos_pendientes), BATCH_SIZE):
        lote_txt  = textos_pendientes[i:  i + BATCH_SIZE]

        for txt in lote_txt:
            # Procesamos el texto Ãºnico
            txt_safe = txt.strip() or "no especificado"
            result = llamar_llm_seguro(txt_safe)
            
            if all(v == -1 for v in result.values()):
                n_errores += 1
                
            nuevas_filas.append({"cc_clean": txt, **result})
            time.sleep(PAUSA_REQ) 

        # Guardado en disco por lote
        df_nuevas = pd.DataFrame(nuevas_filas)
        df_cache_actualizado = pd.concat([df_cache, df_nuevas], ignore_index=True).drop_duplicates("cc_clean")
        df_cache_actualizado.to_parquet(CACHE_PATH, index=False)

        # Progreso
        n_procesados = i + len(lote_txt)
        pct = n_procesados / len(textos_pendientes) * 100
        elapsed = time.time() - t_inicio
        eta_s = (elapsed / max(n_procesados, 1)) * (len(textos_pendientes) - n_procesados)
        print(f"  [{pct:5.1f}%] {n_procesados:>6,}/{len(textos_pendientes):,} "
              f"| Errores: {n_errores} "
              f"| ETA: {eta_s/60:.1f} min", end="\r")

        if i + BATCH_SIZE < len(textos_pendientes):
            time.sleep(PAUSA_LOTE)

    print(f"\n\n  âœ… LLM finalizado en {(time.time()-t_inicio)/60:.1f} min")

# â”€â”€ 5. Reconstruir el Dataset Global (Merge) â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("\n  Reconstruyendo dataset global (418k filas)...")
df_diccionario = pd.read_parquet(CACHE_PATH)

# Cruzamos los resultados del diccionario con el dataset gigante
df_llm_features = df_all.merge(df_diccionario, on="cc_clean", how="left")

# Ordenamos para mantener el Ã­ndice original exacto
df_llm_features["idx"] = df_llm_features["idx"].astype(int)
df_llm_features = df_llm_features.sort_values("idx").reset_index(drop=True)

print(f"  ðŸ“Š Shape final: {df_llm_features.shape} (Debe ser 418100, N)")
print("\nâœ… CELDA 4 OK")
```

```python
# =========================================================
# CELDA 5 â€” ValidaciÃ³n estadÃ­stica de las features LLM
# Â¿CuÃ¡nto separan el acuity 4 del 5? Â¿Y el 3 del 2?
# Si el LLM anotÃ³ bien, deberÃ­amos ver diferencias brutales.
# =========================================================

from scipy.stats import kruskal
from sklearn.metrics import mutual_info_score

# â”€â”€ Reconstruir df_train con features LLM â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# El df_llm_features tiene idx global (train âˆª test)
# Los Ã­ndices del train son los primeros len(df_train_raw)
n_train = len(df_train_raw)

df_llm_train = (
    df_llm_features[df_llm_features["idx"] < n_train]
    .reset_index(drop=True)
)

assert len(df_llm_train) == len(df_train_raw), (
    f"DesalineaciÃ³n: {len(df_llm_train)} vs {len(df_train_raw)}"
)

df_llm_train["acuity"] = df_train_raw["acuity"].values

# â”€â”€ EstadÃ­sticas por nivel de acuity â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("=" * 72)
print("  MEDIA DE CADA FEATURE POR NIVEL DE ACUITY (train set)")
print("  Valores -1 = parse error del LLM (se excluyen del anÃ¡lisis)")
print("=" * 72)

features_binarias = [c for c in CAMPOS_ESPERADOS if c != "urgencia_llm"]
todas_features    = CAMPOS_ESPERADOS  # incluye urgencia_llm

df_valido = df_llm_train[df_llm_train[features_binarias[0]] != -1].copy()
pct_valido = len(df_valido) / len(df_llm_train) * 100
print(f"\n  Filas con parseo vÃ¡lido: {len(df_valido):,} ({pct_valido:.1f}%)\n")

# Tabla de medias por acuity
tabla = (
    df_valido.groupby("acuity")[todas_features]
    .mean()
    .round(3)
)
print(tabla.to_string())

# â”€â”€ Test de Kruskal-Wallis para cada feature â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print(f"\n{'='*72}")
print("  TEST KRUSKAL-WALLIS (Â¿difieren significativamente entre acuities?)")
print(f"{'='*72}")
print(f"  {'Feature':<35} {'H-stat':>10}  {'p-value':>12}  Significativa")

for feat in todas_features:
    grupos = [
        df_valido[df_valido["acuity"] == nivel][feat].values
        for nivel in sorted(df_valido["acuity"].unique())
    ]
    try:
        stat, pval = kruskal(*grupos)
        sig = "âœ… SÃ" if pval < 0.001 else ("âš ï¸  marginal" if pval < 0.05 else "âŒ NO")
        print(f"  {feat:<35} {stat:>10.1f}  {pval:>12.2e}  {sig}")
    except Exception:
        print(f"  {feat:<35}  {'ERROR':>10}")

# â”€â”€ InformaciÃ³n mutua con acuity â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print(f"\n{'='*72}")
print("  INFORMACIÃ“N MUTUA con acuity (mayor = mÃ¡s predictivo)")
print(f"{'='*72}")

y_vals = df_valido["acuity"].values

mi_scores = {}
for feat in todas_features:
    x_vals = df_valido[feat].values
    # mutual_info_score requiere valores discretos
    mi = mutual_info_score(y_vals, x_vals)
    mi_scores[feat] = mi

for feat, mi in sorted(mi_scores.items(), key=lambda x: -x[1]):
    barra = "â–ˆ" * int(mi * 200)
    print(f"  {feat:<35} {mi:.5f}  {barra}")

# â”€â”€ Tabla cruzada: acuity 4 vs 5 en es_tramite_administrativo â”€â”€
print(f"\n{'='*72}")
print("  CROSSTAB: es_tramite_administrativo Ã— acuity (4 y 5 exclusivamente)")
print(f"{'='*72}")

mask_45 = df_valido["acuity"].isin([4, 5])
ct = pd.crosstab(
    df_valido[mask_45]["acuity"],
    df_valido[mask_45]["es_tramite_administrativo"],
    normalize="index",
).round(3) * 100
print(ct.to_string())
print("\n  (% por fila. Acuity 5 deberÃ­a tener >> % en columna 1)")

# â”€â”€ DistribuciÃ³n de urgencia_llm vs acuity real â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print(f"\n{'='*72}")
print("  urgencia_llm asignada por el LLM vs acuity real")
print(f"{'='*72}")
ct2 = pd.crosstab(
    df_valido["acuity"],
    df_valido["urgencia_llm"].clip(1, 5),
    normalize="index",
).round(3) * 100
print(ct2.to_string())
print("\n  (% por fila. Diagonal dominante = el LLM acierta en la direcciÃ³n correcta)")

# â”€â”€ Error de parseo por acuity â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print(f"\n{'='*72}")
print("  TASA DE ERROR DE PARSEO POR ACUITY")
print(f"{'='*72}")
err_mask = df_llm_train[features_binarias[0]] == -1
err_by_acuity = (
    df_llm_train.assign(parse_error=err_mask.astype(int))
    .groupby("acuity")["parse_error"]
    .mean() * 100
)
print(err_by_acuity.round(2).to_string())

print(f"\nâœ… CELDA 5 OK â€” Pega el output completo para diseÃ±ar el paso de integraciÃ³n con LightGBM")
```
